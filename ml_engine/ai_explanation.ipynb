# AI / LLM Narrative Explanation Generator
Generates human-readable narrative explanations and root-cause summaries for flagged data anomalies.

```python
import json
import os

def generate_ai_narrative(anomaly_summary):
    total_anomalies = anomaly_summary.get('total_anomalies', 0)
    quality_score = anomaly_summary.get('quality_score', 100)
    
    explanation = f"Dataset Quality Rating is {quality_score}/100. "
    if total_anomalies > 0:
        explanation += f"Identified {total_anomalies} records requiring attention due to statistical variance or business rule mismatches."
    else:
        explanation += "No critical anomalies detected. Dataset is clean and ready for analytical processing."
        
    return {
        "summary_narrative": explanation,
        "recommended_action": "Review flagged records in the Validation dashboard before ingestion."
    }

sample_summary = {"total_anomalies": 12, "quality_score": 88.5}
print(json.dumps(generate_ai_narrative(sample_summary), indent=2))
```

# Statistical Outlier Detection
Uses Z-Score and Interquartile Range (IQR) to detect statistical anomalies in numeric features.

```python
import pandas as pd
import numpy as np
from scipy import stats
import os

def detect_zscore_outliers(df, threshold=3.0):
    outliers = {}
    num_cols = df.select_dtypes(include=[np.number]).columns
    for col in num_cols:
        col_data = df[col].dropna()
        if len(col_data) > 0:
            z_scores = np.abs(stats.zscore(col_data))
            outlier_mask = z_scores > threshold
            outliers[col] = int(np.sum(outlier_mask))
    return outliers

def detect_iqr_outliers(df, factor=1.5):
    outliers = {}
    num_cols = df.select_dtypes(include=[np.number]).columns
    for col in num_cols:
        col_data = df[col].dropna()
        if len(col_data) > 0:
            q25, q75 = np.percentile(col_data, [25, 75])
            iqr = q75 - q25
            lower_bound = q25 - (factor * iqr)
            upper_bound = q75 + (factor * iqr)
            outliers[col] = int(((col_data < lower_bound) | (col_data > upper_bound)).sum())
    return outliers

DATA_PATH = os.path.join('data', 'dataset_ecommerce_transactions_data.csv')
df = pd.read_csv(DATA_PATH) if os.path.exists(DATA_PATH) else pd.DataFrame()
z_res = detect_zscore_outliers(df)
iqr_res = detect_iqr_outliers(df)
print("Z-Score Outliers:", z_res)
print("IQR Outliers:", iqr_res)
```

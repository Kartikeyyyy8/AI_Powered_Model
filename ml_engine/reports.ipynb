# Automated Report Generator
Generates executive summary PDF, PowerPoint (PPT), and Excel audit reports.

```python
import os
import pandas as pd

def export_excel_report(df, output_path='reports/audit_report.xlsx'):
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
        df.to_excel(writer, sheet_name='Cleaned_Data', index=False)
    print(f"Excel report saved to {output_path}")
    return output_path

DATA_PATH = os.path.join('data', 'dataset_ecommerce_transactions_data.csv')
df = pd.read_csv(DATA_PATH) if os.path.exists(DATA_PATH) else pd.DataFrame()
export_excel_report(df.head(100))
```

# Data Validation Pipeline
Comprehensive checks for missing values, duplicates, datatypes, dates, prices, quantities, payment methods, status, and regex patterns.

```python
import pandas as pd
import numpy as np
import re
import json
import os

DATA_PATH = os.path.join('data', 'dataset_ecommerce_transactions_data.csv')
print(f"Loading data from {DATA_PATH}")

df = pd.read_csv(DATA_PATH) if os.path.exists(DATA_PATH) else pd.DataFrame()
df.head()
```

## 1. Null and Duplicate Checks

```python
def validate_nulls_and_duplicates(df):
    null_counts = df.isnull().sum().to_dict()
    duplicate_count = int(df.duplicated().sum())
    return {"null_counts": null_counts, "duplicate_rows": duplicate_count}

validation_summary = validate_nulls_and_duplicates(df)
print(json.dumps(validation_summary, indent=2))
```

## 2. Schema and Datatype Validation

```python
def validate_dtypes_and_patterns(df):
    results = {}
    for col in df.columns:
        results[col] = {
            "dtype": str(df[col].dtype),
            "unique_values": int(df[col].nunique()),
            "missing_percentage": round(float(df[col].isnull().mean() * 100), 2)
        }
    return results

dtype_summary = validate_dtypes_and_patterns(df)
print(dtype_summary)
```

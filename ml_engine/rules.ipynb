# Business Rule Validation Engine
Checks for negative quantities/prices, invalid future dates, duplicate transaction IDs, and unrecognized product categories.

```python
import pandas as pd
import numpy as np
import os

def run_business_rules(df):
    rule_violations = []
    
    # Rule 1: Negative Price / Quantity
    num_cols = df.select_dtypes(include=[np.number]).columns
    for col in num_cols:
        negatives = df[df[col] < 0]
        if len(negatives) > 0:
            rule_violations.append({
                "rule": f"Negative values in {col}",
                "count": len(negatives),
                "severity": "High"
            })
            
    return pd.DataFrame(rule_violations)

DATA_PATH = os.path.join('data', 'dataset_ecommerce_transactions_data.csv')
df = pd.read_csv(DATA_PATH) if os.path.exists(DATA_PATH) else pd.DataFrame()
violations = run_business_rules(df)
violations
```

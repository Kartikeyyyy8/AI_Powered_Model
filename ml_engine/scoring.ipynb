# Data Quality Scoring Engine
Calculates Dataset Score, Record-level Scores, Completeness Index, and Accuracy Metrics.

```python
import pandas as pd
import numpy as np
import os

def compute_dataset_scores(df):
    if len(df) == 0:
        return {"dataset_score": 0, "completeness_score": 0, "accuracy_score": 0}
        
    completeness = (1 - df.isnull().mean().mean()) * 100
    uniqueness = (1 - (df.duplicated().sum() / len(df))) * 100
    
    overall_score = (completeness * 0.6) + (uniqueness * 0.4)
    return {
        "dataset_score": round(float(overall_score), 2),
        "completeness_score": round(float(completeness), 2),
        "uniqueness_score": round(float(uniqueness), 2)
    }

DATA_PATH = os.path.join('data', 'dataset_ecommerce_transactions_data.csv')
df = pd.read_csv(DATA_PATH) if os.path.exists(DATA_PATH) else pd.DataFrame()
scores = compute_dataset_scores(df)
print("Data Quality Scores:", scores)
```
